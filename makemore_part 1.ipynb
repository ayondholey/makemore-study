```python
g = torch.Generator().manual_seed(2147483647)

for i in range(5):
  out = []
  ix = 0
  while True:
    # Get the row of counts from N corresponding to the current character index `ix`.
    # Add 1 for Laplace smoothing to ensure all probabilities are positive and prevent
    # division by zero if a character has not been followed by any other in the training data.
    # Use `:len(itos)` to ensure we only consider the relevant part of N based on the actual character mapping.
    p_counts = N[ix, :len(itos)].float() + 1

    # Normalize the counts to convert them into a probability distribution
    p = p_counts / p_counts.sum()

    ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[ix])
    if ix == 0: # The index 0 corresponds to the special end token '.'
      break
  print(''.join(out))
```

```python
P = (N+1).float() # model smothening
P /= P.sum(1, keepdims=True)
'''P /= P.sum(1, keepdims=True)"---why this code is used?12:20 AMClaude responded: Two things happening:Two things happening:
N+1 — model smoothing. Every bigram gets count+1 so no probability is ever 0. Without it, unseen bigrams get probability 0 and break the model.
P.sum(1, keepdims=True) — normalize each row so it sums to 1 (valid probability distribution).
The 1 means sum across columns (each row independently). keepdims=True keeps shape (27,1) instead of (27,) so division broadcasts correctly.
'''
```

```python
'''
Quick check — what shape would P.sum(1, keepdims=False) give, and why would dividing fail?

-->
(27,27) divided by (27,) actually does broadcast in PyTorch — but it broadcasts wrong.
It treats (27,) as a row and divides each column by it, instead of each row.
keepdims=True makes shape (27,1) — now it divides each row correctly.
python#
Wrong - divides columns
P / P.sum(1)           # shape (27,)

# Correct - divides rows
P / P.sum(1, keepdims=True)  # shape (27,1)
'''
```

```python
# with different seed and  different W the complete loss will be diferent --that the way of guessing random W and that a bad way of optimizing 'W'
nlls = torch.zeros(5)
for i in range(5):
  # i-th bigram:
  x = xs[i].item() # input character index
  y = ys[i].item() # label character index
  print('--------')
  print(f'bigram example {i+1}: {itos[x]}{itos[y]} (indexes {x},{y})')
  print('input to the neural net:', x)
  print('output probabilities from the neural net:', probs[i])
  print('label (actual next character):', y)
  p = probs[i, y]
  print('probability assigned by the net to the the correct character:', p.item())
  logp = torch.log(p)
  print('log likelihood:', logp.item())
  nll = -logp
  print('negative log likelihood:', nll.item())
  nlls[i] = nll

print('=========')
print('average negative log likelihood, i.e. loss =', nlls.mean().item())
```

```python
# create the dataset
xs, ys = [], []
for w in words:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    xs.append(ix1)
    ys.append(ix2)
xs = torch.tensor(xs)
ys = torch.tensor(ys)
num = xs.nelement()
print('number of examples: ', num)

# initialize the 'network'
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27, 27), generator=g, requires_grad=True)
```

```python
# gradient descent
for k in range(100):

  # forward pass
  xenc = F.one_hot(xs, num_classes=27).float() # input to the network: one-hot encoding
  logits = xenc @ W # predict log-counts
  counts = logits.exp() # counts, equivalent to N
  probs = counts / counts.sum(1, keepdims=True) # probabilities for next character
  loss = -probs[torch.arange(num), ys].log().mean() + 0.01*(W**2).mean() #regularization technique
  print(loss.item())

  # backward pass
  W.grad = None # set to zero the gradient
  loss.backward()

  # update
  W.data += -50 * W.grad #define hyperparameter learning rate based on dataset ..here we needed larger learning rate..do this utill..set iteartions untill we the get loss of Counting Method (2.45.. in this case)
```
